```python
# Create an Age_Group column using simple conditions
df.loc[df["Age"] <= 25, "Age_Group"] = "18-25"
df.loc[(df["Age"] > 25) & (df["Age"] <= 35), "Age_Group"] = "26-35"
df.loc[(df["Age"] > 35) & (df["Age"] <= 45), "Age_Group"] = "36-45"
df.loc[df["Age"] > 45, "Age_Group"] = "46+"

# Count employees in each age group
age_group_count = df["Age_Group"].value_counts()
print("Employees by age group:\n", age_group_count)
print("\nPredominant age group:", age_group_count.idxmax())

plt.figure(figsize=(4, 5))
plt.bar(age_group_count.index, age_group_count.values, color="blue", edgecolor="black")
plt.title("Employee Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Employees")
plt.tight_layout()
plt.show()
```

```python
# 1. Total salary expenditure by team
team_salary = df.groupby("Team")["Salary"].sum()
print("Total salary by team:\n", team_salary)
print("\nTeam with highest salary expenditure:", team_salary.idxmax())
print("Highest team salary amount:", team_salary.max())

# 2. Total salary expenditure by position
position_salary = df.groupby("Position")["Salary"].sum()
print("\nTotal salary by position:\n", position_salary)
print("\nPosition with highest salary expenditure:", position_salary.idxmax())
print("Highest position salary amount:", position_salary.max())

# Bar chart for salary expenditure by team
plt.figure(figsize=(12, 6))
plt.bar(team_salary.index, team_salary.values, color="yellow", edgecolor="black")
plt.title("Total Salary Expenditure by Team")
plt.xlabel("Team")
plt.ylabel("Total Salary")
plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()

# Bar chart for salary expenditure by position
plt.figure(figsize=(4, 5))
plt.bar(position_salary.index, position_salary.values, color="brown", edgecolor="black")
plt.title("Total Salary Expenditure by Position")
plt.xlabel("Position")
plt.ylabel("Total Salary")
plt.tight_layout()
plt.show()
```

```python
# Find correlation between Age and Salary
corr_value = df["Age"].corr(df["Salary"])
print("Correlation between Age and Salary:", corr_value)

# Plot a scatter graph
plt.scatter(df["Age"], df["Salary"], color = "green")
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```
